```python
""" Deleting redundant columns
Renaming the columns 
Dropping duplicates
Cleaning individual columns
Remove the NoN values from the dataset
Check for some more Transformations"""
```

```python
# Filtering the dataset to include only the relevant columns
columns_to_keep = ["date", "adm_level","PCODE", "rfh", "rfh_avg", "r3h", "r3h_avg", "r3q"]
columns_to_drop = ["adm_id", "n_pixels", "r1h", "r1h_avg", "r1q", "version"]
filtered_data = data[columns_to_keep]
print(filtered_data)
```

```python
print(filtered_data.head())
```

```python
# Drop the columns that are not needed for the analysis
data.drop(columns=columns_to_drop , inplace = True, errors="ignore")
```

```python
print(data.shape)
```

```python
# Renaming the columns to more descriptive names
data.rename(columns={"rfh":"current_rainfall", "rfh_avg":"historical_rainfall_avg","r3h":"rolling_3_month_avg","r3q":"seasonal_anomaly_score"}, inplace=True)
print(data)
```

```python
new_column_names = []
for i in data.columns:
    new_column_names.append(i.upper())

print(new_column_names)
```

```python
 # Creating a copy of the original dataset to keep the original data into account in case we need to refer to it later on 
# This allows modifications of the new dataframe without affecting or loosing the original data
new_dataframe_to_modify = data.copy()
# Renaming the columns to uppercase for better reading and consistenty in the new dataframe
new_dataframe_to_modify.columns = new_column_names
print(new_dataframe_to_modify)
```

```python
# Checking for duplicates in the dataset 
new_dataframe_to_modify.duplicated().sum()
```

```python
new_dataframe_to_modify.drop_duplicates()
```

```python
# Remove the NoN values from the dataset
data.isna().sum()
```

```python
data.dropna(inplace=True)
data.isna().sum()
```

```python
# Cleaning individual columns
print(data)
```

```python
# 1. Clean the 'date' Column 
data['date'] = pd.to_datetime(data['date'], errors='coerce')
data['year'] = data['date'].dt.year
data['month'] = data['date'].dt.month
```

```python
# 2. Create your final clean metric 
data['deficit'] = data['current_rainfall'] - data['historical_rainfall_avg']

print("All columns cleaned and verified!")
```
