# Are recent rainfall deficits creating a crisis for Kenyan crops?
- To answer this I used the data from 2022 upto date. Every blue/green bar that points up means that the rain is good and Every red bar pointing down means there's a water deficit

# The Diverging Deflict Chart(Bar Chart)

- What it shows : National monthly rainfall deflict(situation where an area receives significantly less rain/snow tham its normal, expected amount over a specific timeframe) over time.

- Why it's useful: It clearly shows bars dipping below zero in bright red during drought years, giving the judges an instant visual representation of the broken planting calendar.

```python
# Importing libraries
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
data = pd.read_csv(r"C:\Users\PC\Pydata-Capstone-Project\DATA\Kenya_Rainfall data.csv")
```

```python
data.columns
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Compress your data by month for the single year 2022
monthly_2022 = data[data['year'] == 2022].groupby('month')['deficit'].mean().reset_index()

# 2. Set up the plotting window size
plt.figure(figsize=(10, 5))

# 3. Apply dynamic coloring: Red for deficits (below 0), Green for surplus (above 0)
colors = ["red " if x < 0 else "green" for x in monthly_2022['deficit']]

# 4. Generate the bar chart
colors = ["red" if x < 0 else "green" for x in monthly_2022['deficit']]
sns.barplot(data=monthly_2022, x='month', y='deficit', palette=colors)

# 5. Add a bold baseline at zero to make the deficits highly visible
plt.axhline(0, color='black', linewidth=1.5, linestyle='--')

# 6. Label and polish the chart for presentation slides
plt.title("Kenya 2022 Monthly Rainfall Deficit", fontsize=12, fontweight='bold')
plt.xlabel("2022 Calendar Months", fontsize=10)
plt.ylabel("Rainfall Deficit / Surplus (mm)", fontsize=10)

# Replace numeric 1-12 labels with clean month abbreviations
plt.xticks(range(0, 12), ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.savefig("visualizations/monthly_deficit_2022.png", dpi=300) # Saves a high-res slide version
plt.show()
```

Reading 2022 Bars:
- January & February (Green): The year started wet. This gave farmers a false sense of security. They saw the green and assumed it would be a great year to invest money and plant crops.

- October (Red): The second planting season (the "Short Rains") started dry, ruining the late-year crops.

- November & December (Green): Heavy rain finally arrived at the very end of the year, but it was too late to save the food supply for 2022.

QUESTION THREE:
How has the critical March-May planting window shifted geographically and chronologically between 2022 and 2026?

-  March to May is the primary planting season in Kenya. Known as the **"Long Rains"** season, it is the most critical period for agricultural production nationwide.

- I will used a Multi-Line Facet Grid Chart where each line tracks a specific year 


```python
# Filter data for the planting months(March, April and May)
planting_season = data[data["month"].isin([3,4,5])]
# Group by year  and month and calculate the average rainfall for the planting season
annual_planting_rainfall = planting_season.groupby(["year", "month"])["current_rainfall"].mean().reset_index()
historical_planting_avg = planting_season.groupby("month")["historical_rainfall_avg"].mean().reset_index()
# setting plot size
plt.figure(figsize=(12,6))
# plotting the historical average rainfall for the planting season
plt.plot(historical_planting_avg["month"],historical_planting_avg["historical_rainfall_avg"],color="gray",linewidth=4,linestyle="--",label="historical average rainfall")
# plot each year's planting season rainfall using a line plot with markers
sns.lineplot(data=annual_planting_rainfall,x="month",y="current_rainfall",hue="year",marker="o",markersize=8,palette="coolwarm",linewidth=2)
# adding labels and title
plt.title("Kenya Planting Season Rainfall Trends (March-May)",fontsize=14,fontweight="bold")
plt.xlabel("Months",fontsize=12)
plt.ylabel("Rainfall (mm)",fontsize=12)
plt.legend()
plt.show()
```

2022: The Severely Suppressed Year (Dark Blue Line)
The Pattern: 
This line runs completely parallel to the historical average line, but it is drastically shifted downward.

- The Insight: The shape of the season remained traditional (peaking in April), but the overall water volume was highly suppressed.

- The Impact: It represents a nationwide moisture deficit where rains arrived at the right time but simply did not dump enough depth (54 mm) instead of (62 mm) in (April), leading to poor seed germination across PCODEs.


2023: The Mild Relief Year (Light Blue Line)

- The Pattern: A clear step up from 2022. It followed the traditional seasonal curve perfectly, peaking in April at around (86 mm) before dropping down in May.

- The Insight: This year represents a brief "green recovery" period . The rains returned with strong volume, well above the 30-year baseline.

- The Trap: While it looks successful on paper, the sharp drop from April to May highlights how quickly the moisture window opens and shuts, leaving late-planting farmers behind.

 2024: The Hyper-Compressed Deluge (Light Grey Line)
 - The Pattern: This curve is extremely steep and displays a violent, dramatic spike. March was unusually dry (30 mm), but April exploded to a massive peak of (105 mm) nearly double the historical average.
 
 - The Insight: This is the visual signature of climate compression. The same volume of rain that should have been distributed evenly across three months dumped all at once over a single month.
 
 - The Impact: This spike triggers destructive flash floods and severe soil erosion, washing fertilizers out of the soil rather than actually hydrating the crops.
 
  2025: The Delayed, Extended Window (Orange Line)The 
  - Pattern: A flatter, highly unusual curve. March and April stayed stable at around (51 mm) and (72 mm) respectively, but instead of dropping off in May, the rain held steady at (65 mm) far above the normal baseline.
  - The Insight: The planting season experienced a chronological stretch. The rainy season refused to taper off when expected.
  - The Impact: While it provided extended soil moisture, it disrupted the harvest and drying cycles for early-maturing crops, causing storage challenges for farmers.
  
   2026: The Inverted Climate Collapse (Dark Red Line)
   - The Pattern: The most dangerous trend on your chart. This line is completely inverted and slopes downward, cutting straight across your historical baseline.
   
   - The Insight: March saw an unprecedented, unseasonable deluge (77 mm), while April—the historical peak of the long rains—suffered a complete seasonal collapse, plummeting below (50 mm).
   - The Impact: This is the ultimate proof of the "Volatility Trap" . Farmers who planted out of habit in April faced an immediate mid-season drought.

```python
import geopandas as gpd
import matplotlib.pyplot as plt
import pandas as pd

data = pd.read_csv(r"C:\Users\PC\Pydata-Capstone-Project\DATA\Kenya_Rainfall data.csv")
data.columns

```

```python


# Load the official Kenyan spatial map file
spatial_map = gpd.read_file(r"C:\Users\PC\Downloads\ken_admin_boundaries.geojson\ken_admin2.geojson")
spatial_map.head()
```

```python
# Only drops columns where all rows are completely empty
cleaned_spatial_map = spatial_map.dropna(axis=1, how="all")
cleaned_spatial_map.head()
```

```python
cleaned_spatial_map.plot()
plt.title("Kenya Rainfall Deficit by county")
plt.show()
```

```python
# Clean the PCODE columns to ensure a perfect string match
data['PCODE'] = (data['PCODE'].astype(str).str.strip().str.upper())
cleaned_spatial_map['adm2_pcode'] = (cleaned_spatial_map['adm2_pcode'].astype(str).str.strip().str.upper())
```

```python
# Group the critical Mar-May planting season(2022-2026)
planting_season = data[data["month"].isin([3,4,5])].copy()
# Calculate the absolute volumes, expected averages and percentage deficits by PCODE
PCODE_total = planting_season.groupby("PCODE").agg(
    Total_Actual=("current_rainfall", "sum"),
    Total_Expected=("historical_rainfall_avg", "sum"),
).reset_index()




```

```python
# Calculate the missing rainfall colume in mm
PCODE_total["Deficit_mm"] = PCODE_total["Total_Expected"] - PCODE_total["Total_Actual"]
PCODE_total["Deficit_mm"] = PCODE_total["Deficit_mm"].clip(lower=0)# Ensure no negative deficits (surpluses become zero)
```

```python
# Calculate percentage deficit
PCODE_total["Deficit_Percentage"] = (PCODE_total["Deficit_mm"]/PCODE_total["Total_Expected"])*100
```

```python
# merge both files on the cleaned PCODE columns
merged_map = cleaned_spatial_map.merge(PCODE_total, left_on = "adm2_pcode", right_on="PCODE", how="left")
# Fill any missing deficit value with zeroto prevent a flat color crash
merged_map["Deficit_Percentage"] = merged_map["Deficit_Percentage"].fillna(0)
```

```python
# render the final choropleth map
fig, ax = plt.subplots(1, 1, figsize=(12, 8))
merged_map.plot(column="Deficit_Percentage", cmap="Reds", linewidth=0.8, ax=ax, edgecolor="0.8", legend=True)
plt.title("Kenya Planting Season Rainfall Deficit (%) by County (2022-2026)", fontsize=14, fontweight='bold',pad=20)
plt.show()
```

```python
planting_season= data[data["month"].isin([3,4,5])].copy()
planting_season = data[data["month"].isin([3,4,5])].copy()
planting_season["seasonal_anomaly_score"] = (
    planting_season["current_rainfall"] - planting_season["historical_rainfall_avg"]
)

PCODE_whiplash = planting_season.groupby(["PCODE", "year"])["seasonal_anomaly_score"].sum().reset_index()
PCODE_volatility = (
    PCODE_whiplash.groupby("PCODE")["seasonal_anomaly_score"]
    .std()
    .reset_index()
    .rename(columns={"seasonal_anomaly_score": "Whiplash_Score"})
)

merge_map2 = cleaned_spatial_map.merge(PCODE_volatility, left_on="adm2_pcode", right_on="PCODE", how="left")
merge_map2["Whiplash_Score"] = merge_map2["Whiplash_Score"].fillna(0)

fig, ax = plt.subplots(1, 1, figsize=(12, 8))
merge_map2.plot(column="Whiplash_Score", cmap="Oranges", linewidth=0.8, ax=ax, edgecolor="0.8", legend=True)
plt.title("Kenya Planting Season Rainfall Volatility (Whiplash Score) by County (2022-2026)", fontsize=14, fontweight="bold", pad=20)
plt.show()

```
